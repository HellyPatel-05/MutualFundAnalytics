```python
import requests
import pandas as pd
import os

# Create folder if it doesn't exist
os.makedirs("data/raw", exist_ok=True)

# Scheme codes and names
schemes = {
    "119551": "sbi_bluechip",
    "120503": "icici_bluechip",
    "118632": "nippon_large_cap",
    "119092": "axis_bluechip",
    "120841": "kotak_bluechip"
}

for scheme_code, fund_name in schemes.items():

    url = f"https://api.mfapi.in/mf/{scheme_code}"

    try:
        response = requests.get(url)
        response.raise_for_status()

        data = response.json()

        nav_df = pd.DataFrame(data["data"])

        file_path = f"data/raw/{fund_name}_nav.csv"

        nav_df.to_csv(file_path, index=False)

        print(f"✓ Saved: {file_path}")

    except Exception as e:
        print(f"✗ Error fetching {fund_name}: {e}")

print("\nAll NAV files downloaded.")
```

```python
import pandas as pd

fund_master = pd.read_csv("../data/raw/01_fund_master.csv")
nav_history = pd.read_csv("../data/raw/02_nav_history.csv")
```

```python
fund_codes = set(fund_master["amfi_code"])
nav_codes = set(nav_history["amfi_code"])

matching_codes = fund_codes.intersection(nav_codes)
missing_codes = fund_codes - nav_codes

print("Total AMFI Codes in Fund Master:", len(fund_codes))
print("Total AMFI Codes in NAV History:", len(nav_codes))
print("Matching Codes:", len(matching_codes))
print("Missing Codes:", len(missing_codes))
```

```python
if len(missing_codes) > 0:
    print("Missing AMFI Codes:")
    print(sorted(list(missing_codes)))
else:
    print("All AMFI codes from fund_master exist in nav_history.")
```

```python
AMFI Code Validation

Total AMFI Codes in Fund Master: 40
Total AMFI Codes in NAV History: 40
Matching Codes: 40
Missing Codes: 0

Observation:
Most scheme codes were successfully matched between the datasets. AMFI codes serve as the primary key for linking mutual fund scheme information with NAV records.
```
